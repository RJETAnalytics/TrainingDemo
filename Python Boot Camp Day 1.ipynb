Build a simple plot with matplotlib

```python
# Create the scatter plot using matplotlib
plt.scatter(mpg['displ'], mpg['hwy'])

# Add labels and title
plt.xlabel('Engine Displacement')
plt.ylabel('Highway MPG')
plt.title('Engine Displacement vs Highway MPG')

# Show the plot
plt.show()
```

Let's change to a fixed color and make the plot area a bit larger

```python
# Create the scatter plot with a fixed color
plt.figure(figsize=(10, 6))

# Define the facet grid
fgrid = sns.FacetGrid(data=mpg, col="cyl", row="drv")

# Map the scatter plot to each facet
fgrid.map(sns.scatterplot, "displ", "hwy")

# Add labels and title
fgrid.set_axis_labels(xlabel="Engine Displacement", ylabel="Highway MPG")

# Move facets down slightly by using 90% of the plot area so there is room for a title
fgrid.fig.subplots_adjust(top=0.90)

# Set a centered title across all facets
fgrid.fig.suptitle("Engine Displacement vs Highway MPG by Drive Type", fontsize=14, ha='center')

# Show the plot
plt.show()
```

Using the + method to add layers you can include interesting features like a fitted line

```python

# Generate sample scores for 10 assignments/tests in each category
homework_scores = np.array([85, 88, 90, 87, 86, 84, 82, 89, 91, 85])
quiz_scores = np.array([90, 85, 89, 88, 92, 91, 90, 87, 86, 89])
exam_scores = np.array([78, 80, 77, 79, 76, 78, 81, 75, 82, 80])

# Combine into a matrix (each column represents one type of score)
scores_matrix = np.vstack([homework_scores, quiz_scores, exam_scores])

# Define weights
homework_weight = 0.05
quiz_weight = 0.15
exam_weight = 0.8
weights = np.array([[homework_weight], [quiz_weight], [exam_weight]]).T #Transpose from vertical to horizontal

# Compute final weighted scores for each assignment/test
final_scores = np.dot(weights, scores_matrix)  # Dot product for weighted average

print(final_scores)  # Final weighted scores for each of the 10 assignments/tests

print(weights)
```

**Filtering**

**Chaining**

in R we use the pipe %>% but in Python we can use a similar function called chaining using a . (period/dot)

```python
# Method 1: Repetitive lines (without chaining)
CarsDF_6Cyl = mpg.filter(pl.col("cyl")==6)
CarsDF_6Cyl = CarsDF_6Cyl.filter(pl.col("hwy") >= 20)
print(CarsDF_6Cyl.head(10))
```

```python
# Method 2: Method chaining (equivalent to pipes in R)
CarsDF_6Cyl = (
    mpg
    .filter(pl.col("cyl")==6)
    .filter(pl.col("hwy") >= 20)
)
print(CarsDF_6Cyl.head(10))
```

```python
# Method 3: Logical operators in method chaining
CarsDF_6Cyl = mpg.filter((pl.col("cyl") == 6) & ~(pl.col("hwy") >= 20)) # Equivalent to `filter(cyl == 6 & !hwy >= 20)`
#CarsDF_6Cyl = mpg.filter((pl.col("cyl") == 6) & (pl.col("hwy") >= 20)) # Equivalent to `filter(cyl == 6 | hwy >= 20)`
print(CarsDF_6Cyl.head(10))
```

```python
# Filtering the dataset
CarsDF_6Cyl = mpg.filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))

# Sorting: First by 'cyl' (ascending), then by 'hwy' (descending)
CarsDF_6Cyl = CarsDF_6Cyl.sort(
    by=["cyl", "hwy"],
    descending=[False, True])


# Display the sorted DataFrame
print(CarsDF_6Cyl)
```

```python
# Chained Version using () around the whole dataset like R's %>%
CarsDF_6Cyl_Chained = (mpg
    .filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))
    .sort(by=["cyl", "hwy"], descending=[False, True])
)

# Display the sorted DataFrame
print(CarsDF_6Cyl_Chained)
```

**Selecting and de-selecting columns**

```python
# Method 1: Selecting specific columns
CarsDF_6Cyl = (mpg
    .filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))
    .sort(by=["cyl", "hwy"], descending=[False, True])
    .select(['hwy', 'cyl', 'drv'])  # Selecting only these columns
)
print(CarsDF_6Cyl)
```

```python
# Method 2: Deselecting columns
CarsDF_6Cyl = (mpg
    .filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))
    .sort(by=["cyl", "hwy"], descending=[False, True])
    .drop(['trans', 'fl'])  # Dropping unwanted columns
)
print(CarsDF_6Cyl)
```

**Select Helpers**

Python and Pandas doesn't have a something equivilant to dplyrs contains() and other helpers but we can get the same results by using string matching on the column names

```python
import polars.selectors as cs

CarsDF_6Cyl = (
    mpg
    .filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))
    .sort(by=["cyl", "hwy"], descending=[False, True])
    # Select columns that **contain "m" or "c"** in their names
    .select(cs.contains("m") | cs.contains("c"))
)

print(CarsDF_6Cyl.head())
```

**Relocating Columns**

```python
# Method 1: Move 'class' and 'drv' to the front while keeping everything else
CarsDF = (mpg
  .select("class", "drv", pl.all().exclude(["class", "drv"]))
)
print(CarsDF.head())
```

```python
# Method 2: Relocate 'class' to the beginning
CarsDF = (mpg
  .select("class", pl.all().exclude(["class"]))
)
print(CarsDF.head())
```

**Renaming Columns**

```python
# Apply filtering, sorting, column selection, and renaming—all in one chain
CarsDF_6Cyl = (
    mpg
    .filter((pl.col("cyl") == 6) | (pl.col("hwy") >= 20))
    .sort(by=["cyl", "hwy"], descending=[False, True])
    .select(cs.contains("m") | cs.contains("c"))
    .rename({"cty": "CityMiles",
             "class": "VehicleClass",
             "cyl": "Cylinders"})
)

print(CarsDF_6Cyl.head())
```

**Groups and Aggregation**

```python
# Basic aggregation
# Group by 'cyl' and summarize with count & mean
#AGGCarsDF = (
#    mpg
#    .groupby('cyl', as_index=False)  # Group by cylinder count
#    .agg(ModelCount=('model', 'count'), AvgMPG=('hwy', 'mean'))  # Aggregate functions
#)

AGGCarsDF = (mpg
    .group_by("cyl")  # Group by cylinder count
    .agg(ModelCount = pl.col("model").count(),
         AvgMPG = pl.col("hwy").mean())
)

print(AGGCarsDF)
```
